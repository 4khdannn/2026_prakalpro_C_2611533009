## 2.1 Dasar Teori
Pengenalan tentang Nested dan Perulangan
 1. Perulangan

Perulangan atau <i>looping</i> merupakan salah satu konsep dasar dalam pemrograman yang digunakan untuk menjalankan suatu instruksi atau sekumpulan instruksi secara berulang. Perulangan memungkinkan sebuah proses dilakukan beberapa kali tanpa harus menuliskan instruksi yang sama secara berulang. Penggunaan perulangan dapat membuat program menjadi lebih sederhana, efisien, dan mudah dikelola.
<br>
Perulangan bekerja berdasarkan kondisi atau jumlah pengulangan yang telah ditentukan. Selama kondisi yang menjadi dasar perulangan masih terpenuhi, proses akan terus dilakukan. Ketika kondisi tersebut tidak lagi terpenuhi, proses perulangan akan berhenti dan program dapat melanjutkan ke instruksi berikutnya. Oleh karena itu, pemahaman terhadap kondisi perulangan sangat penting agar proses dapat berjalan sesuai dengan tujuan yang diinginkan.
<br>

 2. Jenis-Jenis Perulangan

Dalam pemrograman, perulangan secara umum dapat dilakukan berdasarkan jumlah pengulangan yang sudah diketahui maupun berdasarkan kondisi tertentu. Beberapa bentuk perulangan yang umum digunakan dalam Python adalah <i>for</i> dan <i>while</i>
<br>
Perulangan <i>for</i> digunakan ketika proses pengulangan dilakukan berdasarkan suatu urutan data atau jumlah pengulangan yang telah ditentukan. Perulangan ini banyak digunakan untuk mengakses setiap elemen dalam suatu kumpulan data, seperti daftar, teks, atau rentang angka.
<br>
Perulangan <i>while</i> digunakan ketika pengulangan bergantung pada suatu kondisi tertentu. Proses akan terus dilakukan selama kondisi tersebut bernilai benar. Perulangan ini sesuai digunakan ketika jumlah pengulangan belum diketahui secara pasti dan proses harus berhenti ketika kondisi tertentu telah terpenuhi.
<br>

 3. Nested

<i>Nested</i> merupakan istilah yang digunakan untuk menggambarkan suatu struktur yang berada di dalam struktur lainnya. Dalam pemrograman, konsep <i>nested</i> dapat diterapkan pada beberapa struktur, seperti percabangan maupun perulangan. Struktur yang berada di dalam struktur lainnya akan dijalankan mengikuti aturan dari struktur yang membungkusnya.
<br>
Salah satu bentuk penerapan <i>nested</i> adalah <i>nested loop</i>. <i>Nested loop</i> merupakan perulangan yang berada di dalam perulangan lainnya. Perulangan bagian dalam akan dijalankan sebagai bagian dari proses perulangan bagian luar. Dengan struktur tersebut, sebuah proses dapat dilakukan secara berulang dalam beberapa tingkatan.
<br>

 4. Nested Loop

<i>Nested loop</i> banyak digunakan untuk menyelesaikan permasalahan yang memiliki struktur data bertingkat. Contohnya adalah pengolahan data dalam bentuk baris dan kolom, pembuatan pola, tabel, matriks, serta pengelompokan data. Perulangan luar biasanya digunakan untuk mengatur kelompok atau tingkat pertama, sedangkan perulangan dalam digunakan untuk memproses data pada setiap kelompok tersebut.
<br>
Dalam penerapannya, jumlah proses pada perulangan dalam dapat bergantung pada setiap proses perulangan luar. Oleh karena itu, semakin banyak tingkat perulangan yang digunakan, semakin kompleks pula proses yang harus diperhatikan. Penggunaan <i>nested loop</i> harus dilakukan secara tepat agar program tidak melakukan proses yang tidak diperlukan.
<br>

 5. Hubungan Nested dan Perulangan

Nested dan perulangan memiliki hubungan yang erat karena konsep <i>nested</i> dapat digunakan untuk membuat struktur perulangan yang bertingkat. Dengan adanya perulangan yang berada di dalam perulangan lainnya, program dapat memproses data secara lebih terstruktur berdasarkan tingkatan tertentu.
<br>
Pemahaman mengenai nested dan perulangan penting dalam pembelajaran algoritma karena kedua konsep tersebut dapat membantu menyelesaikan permasalahan yang membutuhkan proses berulang dan terstruktur. Dengan memahami cara kerja perulangan dan struktur bersarang, seseorang dapat menyusun algoritma yang lebih sistematis serta menentukan alur proses program dengan lebih tepat.
<br>

 6. Manfaat Perulangan dan Nested

Perulangan dapat mengurangi pengulangan penulisan instruksi sehingga program menjadi lebih ringkas dan efisien. Sementara itu, nested dapat membantu mengatur proses yang memiliki lebih dari satu tingkatan pengolahan. Kombinasi keduanya dapat digunakan untuk menangani permasalahan yang memiliki banyak data dan pola pengolahan yang berulang.
<br>
Dengan memahami konsep dasar perulangan dan nested, mahasiswa dapat mengembangkan kemampuan dalam menyusun algoritma, memahami alur program, mengolah data secara berulang, serta menyelesaikan permasalahan yang lebih kompleks. Konsep ini juga menjadi dasar untuk mempelajari struktur pemrograman lainnya pada tahap pembelajaran selanjutnya.

## 2.2 Kode Program

### 2.2.1 Program "Perulangan_For1_2611533009"
</div>

```python
#buat program perulangan for dalam python
#program ini menggunakan fungsi input

ulang_3009 = int(input("Masukkan jumlah perulangan: "))

for i_3009 in range(ulang_3009):
    print(f"Perulangan ke-{i_3009}")
```

<div style="font-family:'Times New Roman', Times, serif; font-size: 18px;">

Penjelasan:

1. Input Jumlah Perulangan<br>

2. Penyimpanan Nilai Input<br>
   Nilai 6 disimpan sebagai jumlah perulangan. Karena input diubah menjadi bilangan bulat, nilai tersebut dapat digunakan untuk menentukan batas perulangan.<br>
3. Proses Perulangan<br>
   Perulangan menggunakan fungsi range() dengan jumlah perulangan sebanyak 6 kali. Fungsi tersebut menghasilkan urutan angka mulai dari 0 sampai 5, sehingga terdapat enam nilai yang digunakan dalam proses perulangan.<br>
4. Variabel Penghitung<br>
   Variabel i_3009 berfungsi sebagai penghitung atau penanda pada setiap proses perulangan. Nilainya berubah secara berurutan, yaitu 0, 1, 2, 3, 4, dan 5.<br>
5. Menampilkan Hasil Perulangan<br>
   Pada setiap perulangan, program menampilkan tulisan "Perulangan ke-" yang diikuti dengan nilai dari variabel penghitung. Oleh karena itu, hasil yang ditampilkan dimulai dari perulangan ke-0 hingga perulangan ke-5.<br>

```python
#buat program perulangan for dalam python
#program ini menggunakan fungsi input

batas_3009 = int(input("Masukkan nilai batas: "))
for line_3009 in range(1, batas_3009+1):
    for j_3009 in range(1, (-1 * line_3009 + batas_3009) +1):
        print(".", end=" ")
    print(line_3009)
```

### 2.2.6 Program "Nested_For3_2611533009"
</div>

```python
#buat program perulangan for dalam python
#program ini menggunakan fungsi input

batas_3009 = int(input("Masukkan nilai batas: "))
for i_3009 in range(batas_3009+1):
    for j_3009 in range(batas_3009 + 1):
        print(i_3009 + j_3009, end = "")
    print()
```

<div style="font-family:'Times New Roman', Times, serif; font-size: 18px;">

Penjelasan:

1. Input Nilai Batas<br>
   Program meminta pengguna untuk memasukkan nilai batas. Pada contoh ini, pengguna memasukkan nilai 6. Nilai tersebut digunakan untuk menentukan jumlah perulangan yang dilakukan.<br>
2. Perulangan Pertama<br>
   Perulangan pertama menggunakan variabel i_3009 dengan nilai mulai dari 0 sampai 6. Hal ini terjadi karena fungsi range() dimulai dari 0 dan berhenti sebelum nilai batas yang ditentukan.<br>
3. Perulangan Kedua<br>
   Di dalam perulangan pertama terdapat perulangan kedua menggunakan variabel j_3009. Variabel tersebut juga memiliki nilai mulai dari 0 sampai 6.<br>
4. Proses Penjumlahan<br>
   Pada setiap proses perulangan, nilai i_3009 dan j_3009 dijumlahkan. Hasil penjumlahan tersebut kemudian ditampilkan sebagai output.<br>
5. Menampilkan Hasil<br>
   Hasil penjumlahan ditampilkan secara berurutan tanpa spasi dan tanpa berpindah ke baris baru selama perulangan berlangsung karena menggunakan end="".<br>
6. Jumlah Perulangan<br>
   Jika nilai batas yang dimasukkan adalah 6, maka setiap perulangan berjalan dari 0 sampai 6. Perulangan pertama dilakukan sebanyak 7 kali dan perulangan kedua juga dilakukan sebanyak 7 kali pada setiap perulangan pertama.<br>

### 2.2.7 Program "Nested_For4_2611533009"

```python
#buat program perulangan for dalam python
#program ini menggunakan fungsi input

tinggi_3009 = int(input("Masukkan tinggi pola (bilangan genap, misal 10): "))

if tinggi_3009 % 2 != 0:
    print("Tinggi harus bilangan genap!")
else:
    a_3009 = tinggi_3009 
    c_3009 = a_3009
    lebar_3009 = (2 * tinggi_3009) - 2

    for i_3009 in range(1, tinggi_3009 + 1):
        b_3009 = c_3009 + 1

        for j_3009 in range(1, lebar_3009 + 1):

            # baris atas dan bawah

            if i_3009 == 1 or i_3009 == tinggi_3009:
                if j_3009 == 1 or j_3009 == lebar_3009:
                    print("#", end= " ")
                else:
                    print("=", end= " ")
            # Baris isi
            else:
                if j_3009 == 1 or j_3009 == lebar_3009:
                    print("|", end=" ")
                else:
                    if j_3009 == c_3009:
                        print("<", end="")
                    elif j_3009 == b_3009:
                        print(">", end=" ")
                    elif j_3009 == (lebar_3009 - c_3009):
                        print("<", end=" ")
                    elif j_3009 == (lebar_3009 - c_3009 + 1):
                        print(">", end=" ")
                    elif j_3009 > b_3009 and j_3009 < (lebar_3009 - c_3009):
                        print(".", end=" ")
                    else:
                        print(" ", end=" ")

        print()

        # logika asli java

        a_3009 -= 2

        if a_3009 <= 0:
            c_3009 = (-a_3009) + 2
        else:
            c_3009 = a_3009
```
