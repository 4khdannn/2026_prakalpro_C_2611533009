## 1.1 Latar Belakang
Perulangan (looping) adalah salah satu konsep paling mendasar dalam pemrograman. Dengan perulangan, kita tidak perlu menulis instruksi yang sama berkali-kali. Cara yang cukup efektif untuk melatihnya adalah dengan membuat pola bangun di layar konsol, karena kita harus mengatur sendiri posisi setiap karakter, spasi, dan angka di tiap baris.
Pada praktikum Pekan 5 ini, pola yang dibuat adalah Jam Pasir Kristal Palindromik Berbingkai. Polanya lebih rumit dari pola dasar karena ada deret angka yang menurun dan menaik secara simetris di kiri dan kanan poros <*>, jumlah spasi yang berubah di setiap baris, dan bingkai pembatas yang mengelilingi seluruh pola. Semuanya harus dibuat dengan perulangan for bersarang tanpa perkalian string seperti "=" * n. Karena itu, kita perlu benar-benar memahami hubungan antara nomor baris, jumlah spasi, dan urutan angka yang dicetak.

## 2.1 Dasar Teori
Perulangan for menjalankan sebuah blok kode untuk setiap elemen dalam suatu urutan. Di Python, for hampir selalu dipakai bersama range():
1. range(a) menghasilkan bilangan dari 0 sampai a−1.
2. range(a, b) menghasilkan bilangan dari a sampai b−1.
3. range(a, b, -1) menghasilkan bilangan menurun dari a sampai b+1. Misalnya, range(n, 0, -1) menghasilkan n, n−1, ..., 1.

## 2.2 Kode Program

### 2.2.1 Program "nested"
batas_3009 = int(input("masukkan nilai batas: "))
for i_3009 in range(1, batas_3009 + 1):
    for j_3009 in range(1, (-1 * i_3009 + batas_3009) + 1):
        print(",", end="")
    print(i_3009)
